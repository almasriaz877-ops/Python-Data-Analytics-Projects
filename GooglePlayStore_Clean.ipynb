```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\war\Downloads\googleplaystore.csv\googleplaystore.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Category'] = df['Category'].astype('category')
```

```python
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
```

```python
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
```

```python
df['Size'] = df['Size'].replace('Varies with device', np.nan)
df['Size'] = df['Size'].str.replace('M', '')
df['Size'] = df['Size'].str.replace('k', '')
```

```python
df['Size'] = pd.to_numeric(df['Size'], errors='coerce')
```

```python
df['Installs'] = df['Installs'].str.replace(',', '', regex=False)
df['Installs'] = df['Installs'].str.replace('+', '', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')
```

```python
df['Type'] = df['Type'].astype('category')
```

```python
df['Price'] = df['Price'].str.replace('$','', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
```

```python
df['Content Rating'] = df['Content Rating'].astype('category')
```

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'], errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
```

```python
df['Size'] = df['Size'].fillna(df['Size'].median())
```

```python
df['Reviews'] = df['Reviews'].fillna(df['Reviews'].median())
```

```python
df['Installs'] = df['Installs'].fillna(df['Installs'].median())
```

```python
df['Type'] = df['Type'].fillna(df['Type'].mode()[0])
```

```python
df['Price'] = df['Price'].fillna(df['Price'].mode()[0])
```

```python
df['Content Rating'] = df['Content Rating'].fillna(df['Content Rating'].mode()[0])
```

```python
df['Last Updated'] = df['Last Updated'].fillna(df['Last Updated'].mode()[0])
```

```python
df['Current Ver'] = df['Current Ver'].fillna(df['Current Ver'].mode()[0])
```

```python
df['Android Ver'] = df['Android Ver'].fillna(df['Android Ver'].mode()[0])
```

```python
df.info()
```

```python
df['Price'].info()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ','_')
```

```python
df['reviews'] = df['reviews'].astype(int)
```

```python
df['installs'] = df['installs'].astype(int)
```

```python
df['size'] = df['size'].astype('float32')
```

```python
df['rating'] = df['rating'].astype('float32')
```

```python
df = df[df['rating'] <=5]
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
#Rating distribution

sns.histplot(df['rating'], bins=20, kde=True)
plt.title("Rating Distribution")
plt.show()
```

```python
#Installs distribution

sns.histplot(df['installs'], bins=20)
plt.title("Installs Distribution")
plt.show()
```

```python
# Top Category
df['category'].value_counts().head()
```

```python
df['category'].value_counts().head(10).plot(kind='bar')
plt.title("Top Categories")
plt.show()
```

```python
#Free vs Paid Analysis
df['type'].value_counts()
```

```python
df['type'].unique()
```

```python
df = df[df['type'] !=0]
```

```python
df['type'].value_counts()
```

```python
df['type'] = df['type'].astype(str)
```

```python
df['type'] = df['type'].replace('0', 'Free')
```

```python
df['type'].value_counts()
```

```python
sns.countplot(x='type', data=df)
plt.title("Free VS Paid")
plt.show()

              
```

```python
#Price Analysis (Paid Apps Only)
paid_apps = df[df['price'] > 0]
paid_apps['price'].describe()
```

```python
sns.histplot(paid_apps['price'],bins=20)
plt.title("Paid Apps price distribution")
plt.show()
```

```python
#Size vs Rating Relationship
sns.scatterplot(x='price', y='rating', data=df)
plt.title('Size vs Rating Relationship')
plt.show()
```

```python
#Reviews vs Installs Relationship
sns.scatterplot(x='reviews', y='installs', data=df)
plt.title('Reviews vs Installs')
plt.show()
```

```python
#Category-wise Performance
df.groupby('category')['installs'].sum().sort_values(ascending=False).head(10)
```

```python
df.groupby('category')['installs'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title('Top Categories by Installs')
plt.show()
```

```python
#Average Rating per Category
df.groupby('category')['rating'].mean().sort_values(ascending=False).head(10)
```

```python
df.groupby('category')['rating'].mean().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title('Rating Per Category')
plt.show()
```

```python
#Trend Over Time
df['year'] = df['last_updated'].dt.year
df['year'].value_counts().sort_index().plot(kind='line')
plt.title('Apps Released Over Time')
plt.show()
```

```python
#Correlation Analysis
corr = df[['rating', 'reviews', 'installs', 'price', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True)
plt.title('Correlation Heatmap')
plt.show()
```
