```python
df.groupby("Date")["Weekly_Sales"].sum().plot()
```

```python
#Best Performing Stores
```

```python
df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
```

```python
# Top 10 stores by total sales
```

```python
top_10_stores = (df.groupby('Store')['Weekly_Sales'].sum().sort_values(ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt
```

```python
# Bar Chart
plt.figure(figsize=(10,5))
top_10_stores.plot(kind='bar')

plt.title('Top 10 Stores by Total Sales')
plt.xlabel('Store')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.show()
```
