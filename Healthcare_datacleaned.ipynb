```python
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df["date_of_admission"] = pd.to_datetime(df["date_of_admission"], errors="coerce")
```

```python
df["discharge_date"] = pd.to_datetime(df["discharge_date"], errors="coerce")
```

```python
df['admission_year'] = df['date_of_admission'].dt.year
```

```python
df['admission_month'] = df['date_of_admission'].dt.month_name()
```

```python
#Gender vs Medical Condition
pd.crosstab(
    df['gender'],
    df['medical_condition']
)
```

```python
#Medical Condition vs Billing
df.groupby("medical_condition")["billing_amount"].agg(
    ['count', 'mean', 'sum']
).sort_values("mean", ascending=False)
```

```python
#Insurance vs Billing
df.groupby("insurance_provider")["billing_amount"].agg(
    ['count', 'mean', 'sum']
).sort_values("mean", ascending=False)
```

```python
#Admission Type vs Billing
df.groupby("admission_type")["billing_amount"].agg(
    ['count', 'mean', 'sum'])
```

```python
sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap="coolwarm"
)
plt.show()
```
